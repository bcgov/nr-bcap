### Exploration of Permit Modules
This is to export how to break up the current set of APTS permit attributes into a set of modules
to support the one-permit workflow.

Notes/Questions
---------------
- Can we change the language from "Recommend for" to "xxx Review?"
- Need to think about the MAP might work in this pattern
- Currently assuming there is only one module active at a time (although this may not be the case, especially if we treat Inspection, Investigation and Alteration
as separate modules)
- The Document Approach / Review Approach / Submit Outcome / Review Outcome is the BCFMS pattern. Thinking this might work here but needs to be validated.
This isn't the current world pattern, but seems like it might still fit.
- Thinking that we can still use a default set after Permit Application is submitted (Review PA for Completeness -> Issue Permit), but actual modules and
module order would be manually managed by Permitting Arch (or Project Officer).
